## Übung 1: Variablen und Datentypen

Legen Sie Variablen für Servername, Port, Auslastung und Aktivstatus an. Geben Sie jeweils Wert und Typ aus.

```python
server = "web01"
port = 8080
load = 0.42
active = True

for value in [server, port, load, active]:
    print(value, type(value))
```

## Übung 2: Operatoren

Berechnen Sie aus drei Laufzeiten einen Durchschnitt und prüfen Sie, ob dieser unter einem Grenzwert liegt.

```python
times = [1.2, 1.5, 1.1]
avg = sum(times) / len(times)
limit = 1.5
print(avg)
print(avg < limit)
```

## Übung 3: Strings und f-Strings

Erzeugen Sie aus Variablen eine Statusmeldung wie `web01: Port 8080, aktiv=True`.

```python
name = "web01"
port = 8080
active = True
print(f"{name}: Port {port}, aktiv={active}")
```

## Übung 4: Listen

Erstellen Sie eine Liste mit vier Servernamen. Geben Sie das erste und letzte Element aus, ergänzen Sie einen weiteren Host und zeigen Sie die Anzahl der Hosts.

```python
servers = ["web01", "web02", "db01", "test01"]
print(servers[0])
print(servers[-1])
servers.append("db02")
print(len(servers))
print(servers)
```

## Übung 5: Dictionaries

Erstellen Sie ein Dictionary für einen Server mit `name`, `role`, `port` und `active`. Lesen Sie einzelne Werte aus und ändern Sie den Port.

```python
server = {"name": "web01", "role": "web", "port": 8080, "active": True}
print(server["name"])
print(server["role"])
server["port"] = 9090
print(server)
```

## Übung 6: Verschachtelte Strukturen

Erstellen Sie eine Liste aus mindestens drei Server-Dictionaries und geben Sie einzelne Werte gezielt aus.

```python
servers = [
    {"name": "web01", "role": "web", "port": 8080},
    {"name": "db01", "role": "db", "port": 5432},
    {"name": "test01", "role": "test", "port": 8081},
]
print(servers[0]["name"])
print(servers[1]["port"])
print(servers[2]["role"])
```

## Abschlussübung

Modellieren Sie eine kleine Serverlandschaft als Python-Datenstruktur mit Web-, DB- und Testsystemen. Geben Sie anschließend mindestens fünf gezielte Informationen aus der Struktur aus.

```python
environment = {
    "web": [
        {"name": "web01", "port": 8080},
        {"name": "web02", "port": 8080},
    ],
    "db": [{"name": "db01", "port": 5432}],
    "test": [{"name": "test01", "port": 8081}],
}

print(environment["web"][0]["name"])
print(environment["web"][1]["port"])
print(environment["db"][0]["name"])
print(environment["db"][0]["port"])
print(environment["test"][0])
```

## Erläuterungen und Kontrollfragen

### Übung 1 – Typen
`str` eignet sich für Hostnamen, `int` für Ports, `float` für Auslastungswerte und `bool` für Aktivstatus. `type(wert)` zeigt den tatsächlichen Typ; ein numerisch aussehender String bleibt ohne Konvertierung ein String.

### Übung 2 – Durchschnitt und Grenzwert
Der Durchschnitt ist `sum(laufzeiten) / len(laufzeiten)`; bei einer leeren Liste muss die Division durch null verhindert werden. `<` und `<=` unterscheiden sich am Grenzwert, was in der fachlichen Anforderung geklärt werden sollte.

### Übung 3 – Strings
Ein f-String wie `f"{name}: Port {port}, aktiv={active}"` interpoliert Werte lesbar. Die Ausgabe von `True` und `False` ist die Python-Darstellung boolescher Werte, nicht automatisch eine fachlich übersetzte Statusmeldung.

### Übung 4 – Listen
Listen sind geordnet und veränderbar. Index `0` bezeichnet das erste, `-1` das letzte Element. `append` ergänzt am Ende; Zugriffe auf eine leere Liste mit `[0]` verursachen `IndexError`.

### Übung 5 – Dictionaries
Ein Dictionary ordnet Schlüsseln Werte zu. `server["port"]` verlangt einen vorhandenen Schlüssel, während `server.get("port")` bei Fehlen standardmäßig `None` liefert. Die Wahl hängt davon ab, ob ein fehlendes Feld ein Fehler ist.

### Übung 6 und Abschluss
Eine Liste von Dictionaries modelliert mehrere gleichartig strukturierte Server. Die verschachtelte Struktur nach Rolle ist für kleine Beispiele übersichtlich, aber noch keine Validierung. Bei produktiver Verarbeitung sollten Pflichtfelder und Typen vor dem Zugriff geprüft werden.
