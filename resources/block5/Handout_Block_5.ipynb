## 1. Python im Automatisierungsumfeld

## 2. Variablen und grundlegende Datentypen

Python bestimmt den Typ eines Wertes dynamisch. Wichtige Typen sind `str`, `int`, `float` und `bool`.

````python
name = "web01"
port = 8080
load = 0.42
active = True

for value in (name, port, load, active):
    print(value, "->", type(value).__name__)
````

## 3. Operatoren und Ausdrücke

Operatoren verknüpfen Werte. Beispiele sind `+`, `-`, `*`, `/` sowie Vergleiche wie `==`, `<` und `>`.

````python
port = 8000 + 80
print(port)
print(port == 8080)
print(port > 1024)
````

## 4. Listen

Listen enthalten mehrere Werte in einer definierten Reihenfolge. Der Index beginnt bei `0`.

````python
hosts = ["web01", "web02", "db01"]
print(hosts)
print(hosts[0])
print(len(hosts))
````

## 5. Dictionaries

Dictionaries speichern Werte unter Schlüsseln. Diese Struktur begegnet uns später auch in YAML sehr häufig.

````python
server = {"name": "web01", "port": 8080, "active": True}
print(server["name"])
print(server["port"])
````

## 6. Ein- und Ausgabe

`print()` erzeugt Ausgaben. `input()` liest Benutzereingaben. In Automatisierungsskripten werden Werte häufig statt interaktiv aus Dateien, Argumenten oder anderen Systemen gelesen.

````python
name = "web01"
port = 8080
print(f"Server {name} verwendet Port {port}.")
````

## 7. Python und YAML

Python:

```python
server = {
    "name": "web01",
    "port": 8080
}
```

YAML:

```yaml
server:
  name: web01
  port: 8080
```

Die Syntax unterscheidet sich, das zugrunde liegende Datenmodell ist ähnlich.
