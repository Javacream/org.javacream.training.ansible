## Übung 1: Variablen und Datentypen

Legen Sie Variablen für Servername, Port, Auslastung und Aktivstatus an. Geben Sie jeweils Wert und Typ aus.

```python
server = "web01"
port = 8080
load = 0.42
active = True

for value in [server, port, load, active]:
    print(value, type(value))
```

## Übung 2: Operatoren

Berechnen Sie aus drei Laufzeiten einen Durchschnitt und prüfen Sie, ob dieser unter einem Grenzwert liegt.

```python
times = [1.2, 1.5, 1.1]
avg = sum(times) / len(times)
limit = 1.5
print(avg)
print(avg < limit)
```

## Übung 3: Strings und f-Strings

Erzeugen Sie aus Variablen eine Statusmeldung wie `web01: Port 8080, aktiv=True`.

```python
name = "web01"
port = 8080
active = True
print(f"{name}: Port {port}, aktiv={active}")
```

## Übung 4: Listen

Erstellen Sie eine Liste mit vier Servernamen. Geben Sie das erste und letzte Element aus, ergänzen Sie einen weiteren Host und zeigen Sie die Anzahl der Hosts.

```python
servers = ["web01", "web02", "db01", "test01"]
print(servers[0])
print(servers[-1])
servers.append("db02")
print(len(servers))
print(servers)
```

## Übung 5: Dictionaries

Erstellen Sie ein Dictionary für einen Server mit `name`, `role`, `port` und `active`. Lesen Sie einzelne Werte aus und ändern Sie den Port.

```python
server = {"name": "web01", "role": "web", "port": 8080, "active": True}
print(server["name"])
print(server["role"])
server["port"] = 9090
print(server)
```

## Übung 6: Verschachtelte Strukturen

Erstellen Sie eine Liste aus mindestens drei Server-Dictionaries und geben Sie einzelne Werte gezielt aus.

```python
servers = [
    {"name": "web01", "role": "web", "port": 8080},
    {"name": "db01", "role": "db", "port": 5432},
    {"name": "test01", "role": "test", "port": 8081},
]
print(servers[0]["name"])
print(servers[1]["port"])
print(servers[2]["role"])
```
