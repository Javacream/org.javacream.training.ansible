# Handout – Block 11: Fehlerbehandlung, Debugging und Tests

Dieses Handout erläutert die Lerninhalte des Blocks in verständlicher Form. Ausführbare Beispiele stehen in eigenen Codezellen. Beispiele, die eine vorbereitete Ansible-Trainingsumgebung oder administrative Rechte benötigen, sind so gekennzeichnet, dass sie nicht versehentlich auf einem beliebigen Notebook-System ausgeführt werden.

> Das Handout ergänzt Trainerleitfaden und Übungsnotebook: Es erklärt die Konzepte; die umfangreicheren Aufgaben bleiben im Übungsnotebook.

## 1. Fehler systematisch lesen

Ansible-Ausgaben enthalten wichtige Hinweise: Host, Task, Status und Fehlermeldung.

Grundtypen:

- `ok`
- `changed`
- `failed`
- `unreachable`
- `skipped`

Die erste Debugging-Frage lautet daher: **Welcher Host und welcher Task sind betroffen?**

## 2. `debug`

Mit dem `debug`-Modul können Werte sichtbar gemacht werden:

```yaml
- name: Variable anzeigen
  ansible.builtin.debug:
    var: web_port
```

oder:

```yaml
- ansible.builtin.debug:
    msg: "Host {{ inventory_hostname }} nutzt Port {{ web_port }}"
```

Das ist besonders bei Variablen, Facts und Bedingungen hilfreich.

## 3. Verbosity

Mehr Details auf der Kommandozeile:

```text
ansible-playbook -i inventory.ini site.yml -v
ansible-playbook -i inventory.ini site.yml -vv
ansible-playbook -i inventory.ini site.yml -vvv
```

Mehr Ausgabe ist nicht automatisch besser. Die Verbosity sollte gezielt erhöht werden, wenn die normale Fehlermeldung nicht genügt.

## 4. Fehlerverhalten beeinflussen

`ignore_errors` lässt ein Play trotz eines fehlgeschlagenen Tasks weiterlaufen:

```yaml
ignore_errors: true
```

Das sollte bewusst und sparsam eingesetzt werden, da Fehler sonst leicht verborgen werden.

Mit `failed_when` kann fachlich definiert werden, wann ein Ergebnis als Fehler gilt.

Beispiel:

```yaml
- name: Kommando ausführen
  ansible.builtin.command: some-command
  register: result
  failed_when: result.rc not in [0, 2]
```

`register` speichert das Ergebnis eines Tasks zur weiteren Auswertung.

## 5. Idempotenz testen

Ein wichtiges Testmuster:

```text
1. Playbook ausführen → Änderungen erwartet
2. Playbook erneut ausführen → möglichst changed=0
```

Wenn ein unverändertes System bei jedem Lauf erneut geändert wird, sollte geprüft werden, ob der Task wirklich zustandsorientiert formuliert ist.

## 6. Reproduzierbare Runs

Gute Automatisierung ist:

- nachvollziehbar,
- wiederholbar,
- möglichst idempotent,
- bei Fehlern diagnostizierbar.

### Selbstkontrolle

1. Welche Statuswerte sind für die Fehlersuche wichtig?
2. Wozu dient `debug`?
3. Wann hilft `-vvv`?
4. Warum ist `ignore_errors` mit Vorsicht zu verwenden?
5. Wozu dient `failed_when`?
6. Wie prüft man Idempotenz praktisch?
