# SSH-Demo unter Ubuntu

Benutzer `user0` anlegen, OpenSSH installieren und als systemd-Dienst starten. Die Befehle werden in einem **Ubuntu-Terminal** ausgeführt.

## 1. Benutzer anlegen

````python
%%bash
sudo adduser user0
id user0
getent passwd user0
````

`adduser` richtet Benutzer, Home-Verzeichnis und Passwort ein.

## 2. OpenSSH-Server installieren

````python
%%bash
sudo apt update
sudo apt install openssh-server
````

Der Dienst heißt unter Ubuntu `ssh`.

## 3. SSH-Dienst aktivieren und prüfen

````python
%%bash
sudo systemctl enable --now ssh
systemctl status ssh
sudo ss -tlnp | grep ':22'
````

`enable --now` aktiviert den Autostart und startet den Dienst sofort. Erwartet wird `active (running)`.

## 4. Lokal anmelden

````python
%%bash
ssh user0@localhost
````

Beim ersten Kontakt Host-Key bestätigen und gegebenenfalls Passwort eingeben.

Nach der Anmeldung:

````python
%%bash
whoami
hostname
exit
````

## 5. Von einem anderen Rechner zugreifen

IP-Adresse ermitteln:

````python
%%bash
hostname -I
````

Beispiel (IP-Adresse anpassen):

````python
%%bash
ssh user0@192.168.1.100
````

Falls UFW aktiviert ist:

````python
%%bash
sudo ufw allow OpenSSH
````

## 6. Dienstverwaltung mit systemd

| Befehl | Zweck |
| --- | --- |
| `sudo systemctl start ssh` | Starten |
| `sudo systemctl stop ssh` | Stoppen |
| `sudo systemctl restart ssh` | Neustarten |
| `sudo systemctl enable ssh` | Autostart aktivieren |
| `sudo systemctl disable ssh` | Autostart deaktivieren |
| `systemctl status ssh` | Status |
| `journalctl -u ssh -n 30` | Letzte Logs |

## 7. Optional: Ansible-Zugriff

Inventory-Eintrag (IP-Adresse anpassen):

```ini
[physical]
ubuntu-host ansible_host=192.168.1.100 ansible_user=user0
```

Verbindung von einer erreichbaren Ansible-Control-Node testen:

````python
%%bash
ansible physical -m ping -k
````

`-k` fragt das SSH-Passwort ab. Für dauerhaften Einsatz empfiehlt sich SSH-Key-Authentifizierung.

**Hinweis:** `systemctl` benötigt ein laufendes systemd; bei WSL ist dies gegebenenfalls erst zu aktivieren.
