# Handout – Block 8: Inventories, Variablen und Facts

Dieses Handout erläutert die Lerninhalte des Blocks in verständlicher Form. Ausführbare Beispiele stehen in eigenen Codezellen. Beispiele, die eine vorbereitete Ansible-Trainingsumgebung oder administrative Rechte benötigen, sind so gekennzeichnet, dass sie nicht versehentlich auf einem beliebigen Notebook-System ausgeführt werden.

> Das Handout ergänzt Trainerleitfaden und Übungsnotebook: Es erklärt die Konzepte; die umfangreicheren Aufgaben bleiben im Übungsnotebook.

## 1. Warum Variablen?

Ohne Variablen müssten Werte direkt in Playbooks eingetragen werden. Variablen machen Automatisierung wiederverwendbar.

```yaml
web_port: 8080
```

Verwendung:

```yaml
msg: "Port ist {{ web_port }}"
```

## 2. Inventory-Formate

INI:

```ini
[web]
web01
web02
```

YAML:

```yaml
all:
  children:
    web:
      hosts:
        web01:
        web02:
```

Beide beschreiben Hosts und Gruppen; YAML eignet sich besonders für stärker strukturierte Daten.

## 3. Host- und Gruppenvariablen

Eine übliche Projektstruktur:

```text
inventory.ini
group_vars/
  web.yml
host_vars/
  web01.yml
```

Gruppenvariablen gelten für eine Gruppe, Hostvariablen für einen bestimmten Host.

## 4. Variable Precedence

Eine Variable kann aus mehreren Quellen stammen. Ansible besitzt dafür Prioritätsregeln.

In diesem Block genügt das Grundprinzip:

> Wenn derselbe Variablenname an mehreren Stellen definiert wird, entscheidet die Herkunft darüber, welcher Wert gilt.

Die vollständige Precedence-Liste ist bewusst kein Lernziel.

## 5. Facts

Facts sind von Ansible ermittelte Informationen über Managed Nodes, beispielsweise Betriebssystem, Architektur, Hostname oder Netzwerkschnittstellen.

Beispiel:

```yaml
- debug:
    var: ansible_facts
```

Facts erlauben es, Automatisierung an reale Systemeigenschaften anzupassen.

## 6. Variablen + Facts

```text
eigene Variablen → gewünschte Konfiguration
Facts             → tatsächliche Systemeigenschaften
                         ↓
                  Playbook-Entscheidung
```

Damit kann dasselbe Playbook für unterschiedliche Hosts parametrisiert werden.
