## 1. Warum Ansible?

Ein einzelnes Linux-System lässt sich gut manuell administrieren. Bei vielen Systemen entstehen jedoch wiederkehrende Aufgaben: Pakete installieren, Dateien verteilen, Benutzer verwalten, Dienste konfigurieren oder Deployments durchführen.

Manuelle Administration skaliert schlecht:

```text
Administrator → Server 1
              → Server 2
              → Server 3
              → ...
```

Ansible beschreibt und automatisiert diese Aufgaben zentral. Typische Einsatzfelder sind **Configuration Management, Provisionierung, Deployment** und wiederkehrende Administration.

## 2. Grundarchitektur

Ansible unterscheidet zwei Rollen:

```text
Control Node
   │
   │ SSH
   ├────────► Managed Node 1
   ├────────► Managed Node 2
   └────────► Managed Node 3
```

Der **Control Node** enthält Ansible, Inventory und Automatisierungsdefinitionen. Die **Managed Nodes** sind die Zielsysteme.

Ansible ist grundsätzlich **agentenlos**: Auf den Managed Nodes muss kein dauerhaft laufender Ansible-Agent installiert werden. Für Linux-Ziele wird typischerweise SSH verwendet.

````python
%%bash
echo "SSH-Client auf diesem System:"
ssh -V 2>&1 || true
echo
echo "Python:"
python3 --version 2>&1 || true
````

## 3. Inventory, Hosts und Gruppen

Das Inventory beschreibt, **welche Systeme** Ansible verwalten soll. Hosts können zu Gruppen zusammengefasst werden.

```ini
[web]
web01
web02

[db]
db01
```

Damit kann eine Aktion beispielsweise gezielt für `web`, `db` oder `all` formuliert werden.

Das Inventory ist zunächst nur die Beschreibung der Zielsysteme; die eigentliche Aktion wird durch Ansible-Module bzw. später Playbooks beschrieben.

## 4. Module

Ansible führt Aufgaben über **Module** aus. Ein Modul kapselt eine konkrete Operation, beispielsweise:

- Erreichbarkeit prüfen
- Datei kopieren
- Paket installieren
- Benutzer verwalten
- Dienst starten

Wichtig ist der Unterschied zu einem beliebigen Shell-Kommando: Viele Ansible-Module arbeiten zustandsorientiert. Später kann beispielsweise formuliert werden: „Das Paket soll installiert sein“ statt „Führe diesen Installationsbefehl aus“.

## 5. Shell, Python und Ansible

| Werkzeug | Typische Rolle |
|---|---|
| Shell | vorhandene Programme zu Abläufen verbinden |
| Python | eigene Programmlogik implementieren |
| Ansible | Systemzustände zentral automatisieren |

Ansible kann **ohne selbst geschriebene Python-Skripte** sinnvoll eingesetzt werden. Python spielt technisch im Ansible-Ökosystem eine wichtige Rolle und ist außerdem ein eigenständiges Automatisierungswerkzeug, aber Anwender müssen nicht für jede Ansible-Aufgabe Python programmieren.

## 6. Mentales Modell

```text
Was soll verwaltet werden?       → Inventory
Welche Systeme zusammen?         → Hosts / Gruppen
Welche Aktion?                   → Modul
Wo wird Ansible gestartet?       → Control Node
Wo wirkt die Änderung?           → Managed Node
Wie erfolgt die Verbindung?      → typischerweise SSH
```
