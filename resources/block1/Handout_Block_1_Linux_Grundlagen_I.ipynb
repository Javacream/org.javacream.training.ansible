# 1. Was ist Linux?

Im Alltag bezeichnet **Linux** meist ein komplettes Betriebssystem. Technisch ist Linux zunächst der **Kernel**.

Der Kernel kümmert sich unter anderem um:

- Prozesse,
- Arbeitsspeicher,
- Hardwarezugriffe,
- Dateisysteme,
- grundlegende Systemfunktionen.

Ein vollständiges Linux-System besteht aus mehreren Teilen:

```text
Linux-System
├── Linux-Kernel
├── Systemprogramme
├── Bibliotheken
├── Shell
├── Paketverwaltung
└── Anwendungsprogramme
```

Eine Zusammenstellung dieser Komponenten heißt **Linux-Distribution**.

Beispiele:

- Debian
- Ubuntu
- Red Hat Enterprise Linux
- Fedora
- Rocky Linux
- SUSE Linux Enterprise

Für dieses Seminar ist wichtig: **Ansible automatisiert später Tätigkeiten, die wir zunächst direkt auf Linux kennenlernen.**

# 2. Das Linux-Dateisystem

Linux verwendet einen gemeinsamen hierarchischen Verzeichnisbaum. Der Ausgangspunkt ist das **Root-Verzeichnis** `/`.

```text
/
├── etc
├── home
├── root
├── tmp
├── usr
└── var
```

Wichtige Verzeichnisse:

| Verzeichnis | Bedeutung |
|---|---|
| `/` | Wurzel des Dateisystems |
| `/home` | Home-Verzeichnisse normaler Benutzer |
| `/root` | Home-Verzeichnis von `root` |
| `/etc` | System- und Anwendungskonfiguration |
| `/var` | veränderliche Daten, z. B. Logs |
| `/tmp` | temporäre Dateien |
| `/usr` | Programme, Bibliotheken und weitere Ressourcen |

Anders als unter Windows gibt es nicht für jedes Laufwerk einen separaten Verzeichnisbaum wie `C:\` oder `D:\`.

````python
%%bash
echo "Inhalt des Root-Verzeichnisses:"
ls /

echo
echo "Einige wichtige Verzeichnisse:"
ls -ld /etc /home /tmp /usr /var
````

# 3. Absolute und relative Pfade

Ein **absoluter Pfad** beginnt bei `/`.

Beispiel:

```text
/home/training/projects/demo.txt
```

Ein **relativer Pfad** wird vom aktuellen Arbeitsverzeichnis aus interpretiert.

Besondere Pfadangaben:

| Schreibweise | Bedeutung |
|---|---|
| `.` | aktuelles Verzeichnis |
| `..` | übergeordnetes Verzeichnis |
| `~` | Home-Verzeichnis des aktuellen Benutzers |

````python
%%bash
echo "Aktuelles Arbeitsverzeichnis:"
pwd

echo
echo "Home-Verzeichnis:"
echo "$HOME"
````

Ein relativer Pfad hängt davon ab, **wo Sie sich gerade befinden**.

Die folgende Zelle wechselt nach `/tmp`, dann eine Ebene nach oben und zeigt jeweils das aktuelle Verzeichnis.

````python
%%bash
cd /tmp
echo "Jetzt in:"
pwd

cd ..
echo "Eine Ebene höher:"
pwd
````

### Verständnisfrage

Sie befinden sich in:

```text
/home/training/projects
```

Der Pfad

```text
../data
```

zeigt auf:

```text
/home/training/data
```

# 4. Terminal und Shell

Ein **Terminal** stellt die textorientierte Ein- und Ausgabe bereit.

Die **Shell** interpretiert die eingegebenen Kommandos und startet Programme.

```text
Benutzer
   ↓
Terminal
   ↓
Shell
   ↓
Programme / Linux
```

Eine verbreitete Linux-Shell ist **Bash**.

Ein Kommando besteht typischerweise aus:

```text
Kommando   Option   Argument
   ls        -l       /etc
```

````python
%%bash
ls -l /etc | head
````

# 5. Orientierung in der Shell

Die drei wichtigsten Kommandos zur Navigation sind:

- `pwd` – aktuelles Arbeitsverzeichnis anzeigen
- `ls` – Verzeichnisinhalt anzeigen
- `cd` – Arbeitsverzeichnis wechseln

````python
%%bash
echo "Aktuelles Verzeichnis:"
pwd

echo
echo "Inhalt:"
ls
````

# 6. Dateien und Verzeichnisse verwalten

Für die folgenden Beispiele verwenden wir ein eigenes Übungsverzeichnis:

```text
/tmp/linux-handout-block1
```

So können die Beispiele gefahrlos ausprobiert werden.

````python
%%bash
BASE=/tmp/linux-handout-block1
rm -rf "$BASE"
mkdir -p "$BASE"
echo "Übungsbereich angelegt:"
ls -ld "$BASE"
````

## Verzeichnisse anlegen

`mkdir` erzeugt ein Verzeichnis. Mit `-p` können auch mehrere Ebenen gleichzeitig angelegt werden.

````python
%%bash
BASE=/tmp/linux-handout-block1

mkdir "$BASE/projects"
mkdir -p "$BASE/projects/demo/config"

find "$BASE" -maxdepth 4 -type d | sort
````

## Dateien anlegen

`touch` legt eine Datei an, wenn sie noch nicht existiert.

````python
%%bash
BASE=/tmp/linux-handout-block1

touch "$BASE/demo.txt"
ls -l "$BASE/demo.txt"
````

## Dateien kopieren

`cp` kopiert Dateien.

````python
%%bash
BASE=/tmp/linux-handout-block1

echo "Originalinhalt" > "$BASE/quelle.txt"
cp "$BASE/quelle.txt" "$BASE/kopie.txt"

echo "Quelle:"
cat "$BASE/quelle.txt"

echo
echo "Kopie:"
cat "$BASE/kopie.txt"
````

## Dateien verschieben oder umbenennen

`mv` kann beides:

- Datei an einen anderen Ort verschieben,
- Datei umbenennen.

````python
%%bash
BASE=/tmp/linux-handout-block1

mv "$BASE/kopie.txt" "$BASE/umbenannt.txt"

mkdir -p "$BASE/archive"
mv "$BASE/umbenannt.txt" "$BASE/archive/"

find "$BASE" -maxdepth 2 -type f | sort
````

## Dateien löschen

`rm` löscht Dateien direkt. Normalerweise gibt es dabei keinen Papierkorb.

Die nächste Zelle löscht ausschließlich eine extra erzeugte Testdatei.

````python
%%bash
BASE=/tmp/linux-handout-block1

touch "$BASE/test-delete.txt"
echo "Vor dem Löschen:"
ls -l "$BASE/test-delete.txt"

rm "$BASE/test-delete.txt"

echo
echo "Nach dem Löschen:"
if [ -e "$BASE/test-delete.txt" ]; then
    echo "Datei existiert noch."
else
    echo "Datei wurde gelöscht."
fi
````

# 7. Dateiinhalte betrachten

Für kurze Dateien eignet sich `cat`.

`head` zeigt die ersten Zeilen, `tail` die letzten Zeilen.

`less` eignet sich zum interaktiven Lesen längerer Dateien in einem Terminal.

````python
%%bash
FILE=/tmp/linux-handout-block1/text.txt

printf "Zeile 1\nZeile 2\nZeile 3\nZeile 4\nZeile 5\n" > "$FILE"

echo "cat:"
cat "$FILE"

echo
echo "head:"
head -n 2 "$FILE"

echo
echo "tail:"
tail -n 2 "$FILE"
````

`less` wird in einer normalen Terminal-Sitzung beispielsweise so verwendet:

```text
less /etc/services
```

Mit `q` verlassen Sie `less`.

# 8. Benutzer und Gruppen

Linux ist ein Mehrbenutzersystem.

Ein Benutzer besitzt eine eindeutige Benutzer-ID. Benutzer können Mitglied einer oder mehrerer Gruppen sein.

`whoami` zeigt den aktuellen Benutzer.

`id` zeigt zusätzlich Benutzer-ID, primäre Gruppe und weitere Gruppenzugehörigkeiten.

````python
%%bash
echo "Aktueller Benutzer:"
whoami

echo
echo "Benutzer- und Gruppeninformationen:"
id
````

Systemweite Benutzerinformationen finden sich unter anderem in `/etc/passwd`.

````python
%%bash
head -n 5 /etc/passwd
````

# 9. Das Linux-Rechtekonzept

Mit `ls -l` sehen Sie Eigentümer, Gruppe und Zugriffsrechte.

Beispiel:

```text
-rwxr-xr--
```

Aufteilung:

```text
-  rwx  r-x  r--
│   │    │    └── others
│   │    └─────── group
│   └──────────── owner
└──────────────── Dateityp
```

Bedeutung:

| Recht | Datei | Verzeichnis |
|---|---|---|
| `r` | lesen | Einträge lesen |
| `w` | verändern | Einträge ändern |
| `x` | ausführen | Verzeichnis betreten/durchlaufen |

````python
%%bash
FILE=/tmp/linux-handout-block1/rechte.txt

echo "Inhalt" > "$FILE"
ls -l "$FILE"
````

# 10. Rechte mit `chmod` verändern

Symbolische Schreibweise:

- `u` = owner/user
- `g` = group
- `o` = others
- `+` = Recht hinzufügen
- `-` = Recht entfernen

````python
%%bash
FILE=/tmp/linux-handout-block1/rechte.txt

chmod u+x "$FILE"
echo "Nach u+x:"
ls -l "$FILE"

chmod o-r "$FILE"
echo
echo "Nach o-r:"
ls -l "$FILE"
````

## Numerische Schreibweise

```text
r = 4
w = 2
x = 1
```

Daraus folgt:

```text
7 = rwx
6 = rw-
5 = r-x
4 = r--
```

Beispiele:

```text
755 = rwxr-xr-x
644 = rw-r--r--
```

````python
%%bash
FILE=/tmp/linux-handout-block1/rechte.txt

chmod 755 "$FILE"
echo "755:"
ls -l "$FILE"

chmod 644 "$FILE"
echo
echo "644:"
ls -l "$FILE"
````

# 11. Eigentümer und Gruppe

Eine Datei besitzt:

- einen Eigentümer,
- eine zugeordnete Gruppe,
- Rechte für owner, group und others.

Die folgende Zelle zeigt Eigentümer und Gruppe einer Übungsdatei.

````python
%%bash
FILE=/tmp/linux-handout-block1/rechte.txt
ls -l "$FILE"
````

`chown` ändert den Eigentümer, `chgrp` die Gruppe.

Im Notebook verwenden wir nur den bereits aktuellen Benutzer und dessen primäre Gruppe.

````python
%%bash
FILE=/tmp/linux-handout-block1/rechte.txt
USER_NAME="$(id -un)"
GROUP_NAME="$(id -gn)"

chown "$USER_NAME" "$FILE"
chgrp "$GROUP_NAME" "$FILE"

ls -l "$FILE"
````

# 12. `root` und `sudo`

`root` ist der administrative Benutzer eines Linux-Systems.

Statt permanent als `root` zu arbeiten, verwendet man administrative Rechte gezielt mit `sudo`.

Allgemeines Muster:

```text
sudo <kommando>
```

Das Prinzip:

```text
normaler Benutzer
       │
       │ sudo
       ▼
administrative Aktion
```

Ob `sudo` in einer konkreten Umgebung erlaubt ist, hängt von der Systemkonfiguration ab.

````python
%%bash
echo "Aktueller Benutzer:"
whoami

echo
echo "sudo verfügbar?"
if command -v sudo >/dev/null 2>&1; then
    echo "Ja: $(command -v sudo)"
else
    echo "Nein"
fi
````
