# Übung 5: Ausgabeumleitung und Pipes

## Aufgabe

Arbeiten Sie wieder mit `server.txt`.

1. Schreiben Sie alle Zeilen mit `production` in eine neue Datei `/tmp/linux_block1/production.txt`.
2. Hängen Sie anschließend alle Zeilen mit `testing` an dieselbe Datei an.
3. Geben Sie die fertige Datei aus.
4. Verwenden Sie eine Pipe, um aus `server.txt` nur Zeilen mit `active` anzuzeigen.
5. Leiten Sie dieses Ergebnis in die Datei `/tmp/linux_block1/active.txt` um.
6. Prüfen Sie den Inhalt von `active.txt`.

Verwenden Sie dabei:

- `>`
- `>>`
- `|`
- `grep`
- `cat`


````python
%%bash
FILE=/tmp/linux_block1/projekt/daten/server.txt
BASE=/tmp/linux_block1

grep "production" "$FILE" > "$BASE/production.txt"
grep "testing" "$FILE" >> "$BASE/production.txt"

echo "production.txt:"
cat "$BASE/production.txt"

echo
echo "Aktive Server:"
cat "$FILE" | grep "active"

cat "$FILE" | grep "active" > "$BASE/active.txt"

echo
echo "active.txt:"
cat "$BASE/active.txt"
````

## Denkfrage

Was ist der Unterschied zwischen `>` und `>>`?

Notieren Sie Ihre Antwort:


**Musterantwort:**  

`>` schreibt die Ausgabe in eine Datei und überschreibt dabei einen bereits vorhandenen Inhalt.  
`>>` hängt die Ausgabe an das Ende einer vorhandenen Datei an. Existiert die Datei noch nicht, wird sie angelegt.

# Übung 6: Benutzer und Gruppen untersuchen

## Aufgabe

1. Ermitteln Sie den aktuell angemeldeten Benutzer.
2. Lassen Sie sich dessen Benutzer-ID und Gruppen anzeigen.
3. Geben Sie die ersten Zeilen von `/etc/passwd` aus.
4. Suchen Sie in `/etc/passwd` nach dem Namen Ihres aktuellen Benutzers.

Verwenden Sie insbesondere:

- `whoami`
- `id`
- `cat` oder `head`
- `grep`


````python
%%bash
USER_NAME=$(whoami)

echo "Aktueller Benutzer:"
whoami

echo
echo "Benutzer- und Gruppeninformationen:"
id

echo
echo "Erste Zeilen aus /etc/passwd:"
head /etc/passwd

echo
echo "Eintrag des aktuellen Benutzers:"
grep "^${USER_NAME}:" /etc/passwd
````

## Denkfragen

- Welche Information liefert `whoami`, die `id` ebenfalls enthält?
- Welche zusätzlichen Informationen liefert `id`?
- Warum enthält `/etc/passwd` trotz seines Namens normalerweise keine Benutzerpasswörter?


### Musterantworten

- `whoami` gibt ausschließlich den Namen des aktuell wirksamen Benutzers aus. Diese Information ist auch in der Ausgabe von `id` enthalten.
- `id` zeigt zusätzlich unter anderem Benutzer-ID, primäre Gruppen-ID sowie weitere Gruppenzugehörigkeiten.
- `/etc/passwd` enthält heute normalerweise keine Passwort-Hashes. Diese befinden sich aus Sicherheitsgründen typischerweise in `/etc/shadow`, das nur privilegiert gelesen werden kann.

- Warum hat das `x`-Recht bei einem Verzeichnis eine andere praktische Bedeutung als bei einer normalen Datei?

# Abschlussübung: Eine kleine Arbeitsumgebung aufbauen

Diese Übung kombiniert die wichtigsten Inhalte des Blocks.

## Ausgangssituation

Für ein kleines Administrationsprojekt soll unter `/tmp/linux_block1/abschluss` eine Arbeitsstruktur vorbereitet werden.

## Anforderungen

Erstellen Sie folgende Struktur:

```text
abschluss/
├── config/
├── data/
├── logs/
└── archive/
```

Führen Sie anschließend folgende Aufgaben durch:

1. Kopieren Sie `projekt/config/app.conf` nach `abschluss/config/`.
2. Kopieren Sie `projekt/daten/server.txt` nach `abschluss/data/`.
3. Erzeugen Sie aus `server.txt` eine Datei `production.txt`, die nur Produktionsserver enthält.
4. Speichern Sie `production.txt` im Verzeichnis `data`.
5. Erzeugen Sie eine Datei `logs/setup.log` und schreiben Sie mindestens zwei selbst gewählte Statusmeldungen hinein.
6. Verwenden Sie dabei einmal `>` und einmal `>>`.
7. Kopieren Sie `production.txt` zusätzlich nach `archive`.
8. Setzen Sie für `abschluss/config/app.conf` die Rechte auf `640`.
9. Setzen Sie für das Verzeichnis `config` die Rechte so, dass nur Eigentümer und Gruppe darauf zugreifen können.
10. Prüfen Sie die vollständige Struktur und die gesetzten Rechte.

## Vorgabe

Versuchen Sie, die Aufgabe möglichst ohne Rückgriff auf konkrete Befehlsvorschläge zu lösen. Entscheiden Sie selbst, welches Werkzeug für den jeweiligen Arbeitsschritt geeignet ist.


````python
%%bash
BASE=/tmp/linux_block1
TARGET="$BASE/abschluss"

mkdir -p "$TARGET"/{config,data,logs,archive}

cp "$BASE/projekt/config/app.conf" "$TARGET/config/"
cp "$BASE/projekt/daten/server.txt" "$TARGET/data/"

grep "production" "$TARGET/data/server.txt" > "$TARGET/data/production.txt"

echo "Setup gestartet" > "$TARGET/logs/setup.log"
echo "Produktionsserver extrahiert" >> "$TARGET/logs/setup.log"

cp "$TARGET/data/production.txt" "$TARGET/archive/"

chmod 640 "$TARGET/config/app.conf"
chmod 750 "$TARGET/config"

echo "Abschlussstruktur:"
find "$TARGET" -maxdepth 2 -print | sort

echo
echo "Rechte:"
ls -ld "$TARGET/config"
ls -l "$TARGET/config/app.conf"
````

````python
%%bash
TARGET=/tmp/linux_block1/abschluss

echo "Verzeichnisse:"
find "$TARGET" -maxdepth 1 -type d -print | sort

echo
echo "production.txt:"
cat "$TARGET/data/production.txt"

echo
echo "setup.log:"
cat "$TARGET/logs/setup.log"

echo
echo "Archiv:"
ls -l "$TARGET/archive/production.txt"

echo
echo "Dateirechte app.conf:"
stat -c '%A %a %n' "$TARGET/config/app.conf"

echo
echo "Verzeichnisrechte config:"
stat -c '%A %a %n' "$TARGET/config"
````

# Reflexion – Musterantworten

1. **Wann ist ein absoluter Pfad sinnvoller als ein relativer Pfad?**  
   Wenn eindeutig und unabhängig vom aktuellen Arbeitsverzeichnis auf eine Datei oder ein Verzeichnis zugegriffen werden soll, etwa in Skripten oder bei administrativen Aufgaben.

2. **Was ist der Unterschied zwischen Kopieren und Verschieben einer Datei?**  
   Beim Kopieren bleibt die Quelldatei erhalten und es entsteht eine zusätzliche Datei. Beim Verschieben wird die Datei an einen anderen Ort übertragen bzw. umbenannt.

3. **Wann würden Sie `cat`, wann `less` und wann `grep` verwenden?**  
   `cat` eignet sich für kurze Dateien oder zum direkten Ausgeben. `less` ist für längere Inhalte geeignet, die interaktiv betrachtet werden sollen. `grep` dient zum gezielten Suchen bzw. Filtern von Text.

4. **Was passiert bei `>` mit einer bereits vorhandenen Zieldatei?**  
   Der bisherige Inhalt wird überschrieben.

5. **Welche drei Benutzerklassen berücksichtigt das klassische Linux-Rechtesystem?**  
   Eigentümer (`user`), Gruppe (`group`) und Andere (`others`).

6. **Warum sollte nicht dauerhaft als `root` gearbeitet werden?**  
   `root` darf praktisch alle Systemänderungen durchführen. Fehler oder unbeabsichtigte Befehle können daher wesentlich größere Auswirkungen haben. Administrative Rechte sollten gezielt und nur bei Bedarf eingesetzt werden.

7. **Welche der heute verwendeten Arbeitsschritte könnten später sinnvoll automatisiert werden?**  
   Beispielsweise das Anlegen von Verzeichnissen, Kopieren von Konfigurationsdateien, Setzen von Berechtigungen, Filtern bzw. Erzeugen von Dateien und die wiederholte Vorbereitung identischer Systemstrukturen.
