# Handout – Block 7: Ansible Playbooks – Deklarative Automatisierung

Dieses Handout erläutert die Lerninhalte des Blocks in verständlicher Form. Ausführbare Beispiele stehen in eigenen Codezellen. Beispiele, die eine vorbereitete Ansible-Trainingsumgebung oder administrative Rechte benötigen, sind so gekennzeichnet, dass sie nicht versehentlich auf einem beliebigen Notebook-System ausgeführt werden.

> Das Handout ergänzt Trainerleitfaden und Übungsnotebook: Es erklärt die Konzepte; die umfangreicheren Aufgaben bleiben im Übungsnotebook.

## 1. Vom Ad-hoc Command zum Playbook

Ad-hoc Commands sind für einzelne Aktionen praktisch. Ein **Playbook** beschreibt wiederholbare Automatisierung in einer Datei.

```text
Ad-hoc:  einzelne unmittelbare Aktion
Playbook: dokumentierter, versionierbarer Ablauf
```

## 2. YAML-Grundlagen

YAML verwendet Einrückung zur Strukturierung. Tabs sollten vermieden werden.

```yaml
- name: Webserver konfigurieren
  hosts: web
  become: true
  tasks:
    - name: Paket installieren
      ansible.builtin.package:
        name: nginx
        state: present
```

Listen beginnen mit `-`, Dictionaries bestehen aus Schlüssel-Wert-Paaren.

## 3. Play, Hosts und Tasks

```text
Playbook
└── Play
    ├── hosts
    ├── become
    └── tasks
        ├── Task 1 → Modul
        ├── Task 2 → Modul
        └── Task 3 → Modul
```

Ein Task erhält idealerweise einen aussagekräftigen `name`.

## 4. Zustandsorientierte Module

Beispiel:

```yaml
- name: Paket sicherstellen
  ansible.builtin.package:
    name: nginx
    state: present
```

Die Aussage lautet nicht „installiere nginx jetzt“, sondern „nginx soll vorhanden sein“. Genau daraus entsteht die Grundlage für **Idempotenz**.

## 5. Playbook ausführen

```text
ansible-playbook -i inventory.ini site.yml
```

Typische Status:

- `ok` – gewünschter Zustand war bereits vorhanden
- `changed` – Ansible musste etwas ändern
- `failed` – Task ist fehlgeschlagen

Ein zweiter Lauf eines sauber idempotenten Playbooks sollte möglichst keine unnötigen Änderungen melden.

## 6. Imperativ und deklarativ

Imperativ:

```text
Führe diese Schritte aus.
```

Deklarativ:

```text
Dieser Zustand soll gelten.
```

Ansible kombiniert beide Sichtweisen, viele Standardmodule sind aber bewusst zustandsorientiert.
