# Übungen – Block 7: Ansible Playbooks

Für die Übungen wird die Ansible-Umgebung aus Block 4 weiterverwendet.

## Musterlösung

## Übung 1: YAML lesen

Überführen Sie eine kleine Liste von Servern und Ports in gültiges YAML.

```yaml
servers:
  - name: web01
    port: 8080
  - name: db01
    port: 5432
```

## Übung 2: Erstes Playbook

Erstellen Sie ein Playbook, das auf allen Hosts den Hostnamen mit dem `command`-Modul ausgibt.

```yaml
- name: Hostnamen ausgeben
  hosts: all
  gather_facts: false
  tasks:
    - name: Hostname ermitteln
      ansible.builtin.command: hostname
```

```bash
ansible-playbook -i inventory.ini playbook.yml
```

## Übung 3: Datei verwalten

Erstellen Sie per `file` ein Verzeichnis und per `copy` eine Datei unter `/tmp/ansible-demo`.

```yaml
- name: Demo-Dateien verwalten
  hosts: all
  tasks:
    - name: Verzeichnis anlegen
      ansible.builtin.file:
        path: /tmp/ansible-demo
        state: directory

    - name: Datei anlegen
      ansible.builtin.copy:
        dest: /tmp/ansible-demo/info.txt
        content: "managed by ansible\n"
```

## Übung 4: Paket verwalten

Stellen Sie mit `package` sicher, dass ein freigegebenes Paket installiert ist.

```yaml
- name: Paket installieren
  hosts: all
  become: true
  tasks:
    - name: Paket vorhanden
      ansible.builtin.package:
        name: tree
        state: present
```

## Übung 5: Dienstzustand

Stellen Sie mit einem geeigneten Modul sicher, dass ein freigegebener Dienst läuft.

```yaml
- name: Dienst verwalten
  hosts: web
  become: true
  tasks:
    - name: Dienst läuft
      ansible.builtin.service:
        name: ssh
        state: started
```

## Übung 6: Idempotenz

Führen Sie dasselbe Playbook zweimal aus und vergleichen Sie `changed`-Werte.

```bash
ansible-playbook -i inventory.ini demo.yml
ansible-playbook -i inventory.ini demo.yml
```

Beim zweiten Lauf sollten zustandsorientierte Tasks möglichst `ok` statt `changed` melden.

## Abschlussübung

Erstellen Sie ein Playbook, das Verzeichnis, Konfigurationsdatei, Paket und Dienstzustand verwaltet.

```yaml
- name: Abschluss Block 7
  hosts: web
  become: true
  tasks:
    - name: Verzeichnis vorhanden
      ansible.builtin.file:
        path: /opt/training
        state: directory

    - name: Konfiguration vorhanden
      ansible.builtin.copy:
        dest: /opt/training/app.conf
        content: "environment=training\n"

    - name: Paket vorhanden
      ansible.builtin.package:
        name: tree
        state: present

    - name: SSH-Dienst läuft
      ansible.builtin.service:
        name: ssh
        state: started
```
