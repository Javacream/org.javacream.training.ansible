# Handout – Block 10: Rollen und Strukturierung größerer Ansible-Projekte

Dieses Handout erläutert die Lerninhalte des Blocks in verständlicher Form. Ausführbare Beispiele stehen in eigenen Codezellen. Beispiele, die eine vorbereitete Ansible-Trainingsumgebung oder administrative Rechte benötigen, sind so gekennzeichnet, dass sie nicht versehentlich auf einem beliebigen Notebook-System ausgeführt werden.

> Das Handout ergänzt Trainerleitfaden und Übungsnotebook: Es erklärt die Konzepte; die umfangreicheren Aufgaben bleiben im Übungsnotebook.

## 1. Warum Rollen?

Ein einzelnes Playbook wird mit wachsendem Umfang schwer wartbar. Rollen teilen Automatisierung in wiederverwendbare, klar strukturierte Bestandteile.

```text
Playbook
├── Rolle webserver
├── Rolle database
└── Rolle monitoring
```

## 2. Rollenstruktur

Typische Struktur:

```text
roles/
└── webserver/
    ├── tasks/
    │   └── main.yml
    ├── handlers/
    │   └── main.yml
    ├── templates/
    ├── files/
    ├── defaults/
    │   └── main.yml
    └── vars/
        └── main.yml
```

Nicht jedes Verzeichnis muss in jeder Rolle vorhanden sein.

## 3. Aufgaben verteilen

- `tasks` – auszuführende Tasks
- `handlers` – Handler
- `templates` – Jinja2-Templates
- `files` – statische Dateien
- `defaults` – gut überschreibbare Standardwerte
- `vars` – rolleninterne Variablen mit höherer Priorität

Eine gute Rolle trennt **Logik, Daten und Dateien/Templates** möglichst sauber.

## 4. Rolle verwenden

```yaml
- name: Webserver
  hosts: web
  become: true
  roles:
    - webserver
```

Das aufrufende Playbook bleibt dadurch kompakt.

## 5. Wiederverwendbarkeit

Eine Rolle sollte nicht unnötig auf einen einzigen Host zugeschnitten sein. Konfigurierbare Werte gehören typischerweise in Variablen.

```text
Rolle + Variablen → wiederverwendbarer Baustein
```
