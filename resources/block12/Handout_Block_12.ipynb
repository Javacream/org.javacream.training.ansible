## 1. Ziel des Praxisprojekts

Der Abschlussblock verbindet die Werkzeuge des Seminars in einem zusammenhängenden Szenario.

Wichtig ist nicht, überall jedes Werkzeug einzusetzen, sondern **begründet das passende Werkzeug zu wählen**.

```text
Linux   → Zielsystem und administrative Grundlagen
Shell   → optional einfache Ablaufsteuerung
Python  → eigene Verarbeitungslogik
Ansible → Konfiguration mehrerer Managed Nodes
```

## 2. Möglicher Projektfluss

```text
Ausgangsdaten
    │
    ▼
Python-Skript
    │ erzeugt/verarbeitet Daten
    ▼
Ansible
    │
    ├── Inventory
    ├── Variablen/Facts
    ├── Rolle
    ├── Paket/Dienst
    ├── Template
    └── Handler
    ▼
konfigurierte Managed Nodes
```

Eine Shell kann optional den Gesamtablauf starten, wenn eine einfache Sequenz genügt.

## 3. Anforderungen an den Ansible-Teil

Das Projekt sollte die zentralen Konzepte zusammenführen:

- mehrere Managed Nodes
- Inventory und Gruppen
- Variablen
- Facts
- Paket- oder Dienstverwaltung
- Template
- Handler
- Rolle
- Idempotenzprüfung
- Fehleranalyse

## 4. Rolle von Python

Python sollte dort eingesetzt werden, wo tatsächlich **eigene Programmlogik** benötigt wird, beispielsweise:

- Daten einlesen und validieren
- Werte berechnen
- strukturierte Daten transformieren
- eine Eingabedatei für den Automatisierungsprozess erzeugen

Python sollte nicht künstlich Aufgaben übernehmen, die ein vorhandenes Ansible-Modul bereits sauber zustandsorientiert löst.

````python
servers = [
    {"name": "web01", "port": 8080},
    {"name": "web02", "port": 8081},
]

for server in servers:
    print(f"{server['name']}:{server['port']}")
````

## 5. Idempotenz als Abnahmekriterium

Das fertige Ansible-Projekt wird mindestens zweimal ausgeführt:

```text
Lauf 1 → notwendige Änderungen
Lauf 2 → keine unnötigen Änderungen
```

Falls der zweite Lauf erneut Änderungen meldet, wird untersucht, welcher Task dafür verantwortlich ist.

## 6. Fehler bewusst analysieren

Ein Praxisprojekt ist auch dann wertvoll, wenn Fehler auftreten. Entscheidend ist ein systematisches Vorgehen:

```text
Welcher Host?
   ↓
Welcher Task?
   ↓
FAILED oder UNREACHABLE?
   ↓
Variablen/Facts korrekt?
   ↓
Rechte / become?
   ↓
Modulargumente?
   ↓
Verbosity / debug
```

## 7. Abschlussreflexion

Für jeden Teil des Projekts sollte begründet werden:

- Warum wurde hier Linux-Wissen benötigt?

- Welche Teile sind wiederverwendbar?
- Ist der Ablauf reproduzierbar und idempotent?
