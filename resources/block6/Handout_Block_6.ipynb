## 1. Von Daten zu Programmlogik

Block 5 behandelte Werte und Datenstrukturen. Jetzt steuern wir den Programmablauf mit Bedingungen, Schleifen und Funktionen.

## 2. Bedingungen

```python
status = "active"
if status == "active":
    print("Dienst ist aktiv")
else:
    print("Dienst ist nicht aktiv")
```

## 3. Schleifen

`for` eignet sich besonders für Listen und Dictionaries.

```python
hosts = ["web01", "web02", "db01"]
for host in hosts:
    print("Prüfe", host)
```

```python
ports = {"web": 80, "ssh": 22}
for service, port in ports.items():
    print(service, port)
```

## 4. Funktionen

Funktionen kapseln wiederverwendbare Logik. Parameter liefern Eingaben, `return` einen Rückgabewert.

```python
def endpoint(host, port):
    return f"{host}:{port}"

print(endpoint("web01", 8080))
```

## 5. Module

Mit `import` wird vorhandene Funktionalität eingebunden. Die Standardbibliothek enthält bereits viele nützliche Module.

```python
from pathlib import Path
p = Path("/tmp")
print(p.exists(), p.is_dir())
```

## 6. Fehlerbehandlung

Exceptions signalisieren Fehlerzustände. `try`/`except` erlaubt eine kontrollierte Reaktion.

```python
try:
    port = int("nicht-eine-zahl")
except ValueError as exc:
    print("Ungültiger Port:", exc)
```

## 7. Dateien

```python
from pathlib import Path
p = Path("/tmp/python-block6.txt")
p.write_text("web01\nweb02\n", encoding="utf-8")
print(p.read_text(encoding="utf-8"))
```

## 8. Externe Programme

Python kann externe Programme starten. Dafür eignet sich `subprocess`. Im Unterschied zu einer einfachen Shell-Sequenz kann Python Ergebnisse strukturiert weiterverarbeiten.

```python
import subprocess
r = subprocess.run(["hostname"], capture_output=True, text=True, check=True)
print(r.stdout.strip())
```

## 9. Python oder Ansible?

Ein Python-Skript ist sinnvoll, wenn eigene Programmlogik oder Datenverarbeitung benötigt wird. Ansible ist häufig besser geeignet, wenn auf Systemen ein gewünschter Zustand hergestellt werden soll.

Ansible-Module sind häufig in Python implementiert; daraus folgt aber **nicht**, dass Ansible-Anwender eigene Python-Module schreiben müssen.
