## Übung 7: JSON

Speichern Sie eine Liste von Server-Dictionaries als JSON und lesen Sie sie wieder ein.

```python
import json
from pathlib import Path
servers=[{"name":"web01","role":"web"},{"name":"db01","role":"db"}]
p=Path("/tmp/servers.json")
p.write_text(json.dumps(servers, indent=2), encoding="utf-8")
loaded=json.loads(p.read_text(encoding="utf-8"))
print(loaded)
```

## Abschlussübung

Lesen Sie eine JSON-Datei mit Serverdaten, validieren Sie Pflichtfelder, zählen Sie Rollen und führen Sie für jeden gültigen Eintrag eine kleine lokale Prüfaktion aus.

```python
import json, subprocess
from pathlib import Path
from collections import Counter

data = [
    {"name":"web01","role":"web"},
    {"name":"db01","role":"db"},
    {"name":"test01","role":"test"},
]
p=Path("/tmp/serverdata.json")
p.write_text(json.dumps(data), encoding="utf-8")

servers=json.loads(p.read_text())
valid=[]
for s in servers:
    if "name" in s and "role" in s:
        valid.append(s)

print(Counter(s["role"] for s in valid))

for s in valid:
    r=subprocess.run(["echo", s["name"]], capture_output=True, text=True)
    print(r.returncode, r.stdout.strip())
```
