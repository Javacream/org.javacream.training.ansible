## 2. Ein erstes Inventory

Mit `ansible-inventory -i inventory.ini --graph` lässt sich die Struktur prüfen. In der Trainingsumgebung wird das bereitgestellte Inventory verwendet.

## 3. Ad-hoc Commands

Grundform:

```text
ansible <host-pattern> -i <inventory> -m <modul> -a "<argumente>"
```

Beispiele aus diesem Block:

```text
ansible all -i inventory.ini -m ping
ansible web -i inventory.ini -m command -a "hostname"
ansible web -i inventory.ini -m copy -a "src=demo.txt dest=/tmp/demo.txt"
```

Ad-hoc Commands sind gut für einzelne, unmittelbare Aktionen. Wiederholbare Abläufe werden später als Playbooks beschrieben.

## 4. `ping` ist kein ICMP-Ping

`ansible.builtin.ping` prüft die Ansible-Verbindung und Modulausführung. Ein erfolgreiches Ergebnis enthält typischerweise `pong`.

Das unterscheidet sich vom Linux-Kommando `ping`, das ICMP verwendet.

## 5. Administrative Aktionen mit `become`

SSH und Rechteerhöhung sind getrennte Vorgänge:

```text
Control Node
   │ SSH als training
   ▼
Managed Node
   │ become / sudo
   ▼
root-Rechte für den Task
```

Kurzform:

```text
-b = --become
-K = --ask-become-pass
```

Merksatz: **SSH bestimmt, als wer ich mich verbinde; `become` bestimmt, mit welchen Rechten der Task ausgeführt wird.**

## 6. Status verstehen

Typische Ergebnisse:

- `SUCCESS` – Ausführung erfolgreich
- `CHANGED` – erfolgreich und Systemzustand verändert
- `FAILED` – Host erreicht, Task fehlgeschlagen
- `UNREACHABLE` – Verbindung zum Host nicht möglich

Diese Statusbegriffe werden bei Playbooks erneut wichtig.
